## Chains in LangChain

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

llm_googleai = ChatGoogleGenerativeAI(model="gemini-flash-latest", temperature=0)
```

## First Chain

```python
# Task-1 : Create a prompt template

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a NBA sports facts generator who uses gen-z slang and emojis."),
    ("human", "{input}")
])
```

```python
# Task-2 : LLM

llm_googleai = ChatGoogleGenerativeAI(model="gemini-flash-latest", temperature=0)
```

```python
# Task-3 : String Parser

str_parser = StrOutputParser()
```

## Manual Invocation

```python
# template = prompt_template.invoke({"input" : "Share a fun NBA sports fact."})
# res = llm_googleai.invoke(template)
# final_result = res.content
```

## Chain Invocation

```python
chain = prompt_template | llm_googleai | str_parser

chain.invoke({"input" : "Share a NBA sports fact."})
```
