## Using Pydantic Models

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

llm_googleai = ChatGoogleGenerativeAI(model="gemini-flash-latest", temperature=0)
```

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup : str = Field(description="The setup of the sports fact")
    prompt : str = Field(description="The prompt of the sports fact")
```

```python
obj = llm_schema(**{"setup": "String", "prompt": "String"
})
obj
```

```python
llm_structured_output = llm_googleai.with_structured_output(llm_schema)
result = llm_structured_output.invoke("Share a fun NBA sports fact.")
result.prompt
```

## Using TypedDict

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup : str
    prompt : str
```

```python
obj = llm_schema_td({"setup" : "Strings", "prompt" : "Strings"})
obj["prompt"]
```

```python
llm_structured_output = llm_googleai.with_structured_output(llm_schema_td)
result = llm_structured_output.invoke("Share a fun NBA sports fact.")
result
```
