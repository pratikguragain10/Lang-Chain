# Messages

```python
from langchain_core.messages import  HumanMessage, SystemMessage, AIMessage
from langchain_google_genai import ChatGoogleGenerativeAI

llm_googleai = ChatGoogleGenerativeAI(model="gemini-flash-latest", temperature=0)

messages = [
    SystemMessage(content="You are a gen-z backend expert who also uses slang and emojis in your responses."),
    HumanMessage(content="Explain Django REST Framework in simple terms")
]

response = llm_googleai.invoke(messages)
print(response.content)
```
